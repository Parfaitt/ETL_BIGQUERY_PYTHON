```python
from google.cloud import bigquery
import os
import warnings
warnings.filterwarnings('ignore')


# ===================== CONFIGURATION =====================
PROJECT_ID = "iron-arbor-432008-r7"
DATASET_ID = "my_dataset"
CSV_FOLDER = "D:\RECO\PROJECT 2026\BIG QUERY PROJECT\DATASET"

# Chaque fichier correspond à une table BigQuery
FILES = [
    "account",
    "card",
    "client",
    "disp",
    "district",
    "loan",
    "orders",
    "trans"
]

# =========================================================


# Création du client BigQuery (authentification via service account ou ADC)
client = bigquery.Client(project=PROJECT_ID)


# Configuration du job de chargement
job_config = bigquery.LoadJobConfig(
    source_format=bigquery.SourceFormat.CSV,  # Format source : CSV
    skip_leading_rows=1,                      # Ignore la première ligne (header)
    autodetect=True,                          # Détection automatique du schéma
    write_disposition="WRITE_TRUNCATE"        # Remplace la table si elle existe
)


# Boucle sur chaque fichier CSV
for table in FILES:

    # Construction du chemin vers le fichier CSV
    csv_path = os.path.join(CSV_FOLDER, f"{table}.csv")

    # Construction de l'identifiant complet de la table BigQuery
    # Format : projet.dataset.table
    table_id = f"{PROJECT_ID}.{DATASET_ID}.{table}"

    # Log du chargement en cours
    print(f"📥 Chargement {csv_path} → {table_id}")

    # Ouverture du fichier CSV en mode binaire
    with open(csv_path, "rb") as f:

        # Lancement du job de chargement vers BigQuery
        job = client.load_table_from_file(
            f,
            table_id,
            job_config=job_config
        )

    # Attente de la fin du job (bloquant)
    job.result()

    # Confirmation du chargement
    print(f"✅ Table {table} créée avec succès")


# Message final après chargement de tous les fichiers
print("🚀 Toutes les tables ont été chargées avec succès")
```
